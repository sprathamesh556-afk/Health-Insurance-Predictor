```python
import chromadb
```

```python
#Ephemeral Client
client= chromadb.Client()
```

```python
collection=client.create_collection(name="News")
```

```python
collection.add(
    documents=[
        "Apple reported its quarterly earnings today.", #chunk1
        "Apple has a lot of Vitamin A" #chunk 2
    ],
    ids=["id1","id2"]
)
```

```python
objects=collection.peek()
objects
```

```python
len(objects['embeddings'][1])
```

```python
results= collection.query(
    query_texts=["Oranges are better than mangoes"],
    n_results=2
)
```

```python
results
```

# Persistent Client

```python
clientp=chromadb.PersistentClient(path="./news_vector_db")
```

```python
from chromadb.utils import embedding_functions

ef= embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-distilroberta-v1"
)
```

```python
collection=clientp.create_collection(
    name="news_V2",
    embedding_function=ef
)
```

```python
collection.add(
    documents=[
        "The Taj Mahal is one of the Seven Wonders of the World.",
        "Cricket is the most popular sport in India.",
        "Bengaluru is known as the Silicon Valley of India.",
        "Electric vehicles reduce carbon emissions."
    ],
    ids=["id19","id20","id21","id22"]
)
```

```python
objects = collection.peek()
len(objects['embeddings'][1])
```

```python
results = collection.query(
    query_texts=["Famous monument in india"],
    n_results=2
)
```

```python
results
```

```python
client=chromadb.Client()

```

```python
from datetime import datetime
collection=client.create_collection(
    name="reviews",
    metadata={
        "description":"Product Reviews",
        "created":str(datetime.now())
    }
)
```

```python
collection = client.get_collection("reviews")
```

```python
collection.add(
    documents=[
        "The delivery was fast and the product quality is excellent!",
        "I was not able to increase TV's brightness so I returned it back",
        "The shoes I ordered were too small. Sizing is inaccurate",
        "Great customer support. Resolved my issue in minutes"
    ],
    ids=["r1","r2","r3","r4"],
    metadatas=[
        {"product_category":"electronics","rating":5},
        {"product_category":"electronics","rating":2},
        {"product_category":"apparel","rating":3},
        {"product_category":"services","rating":4}
    ]
)
```
