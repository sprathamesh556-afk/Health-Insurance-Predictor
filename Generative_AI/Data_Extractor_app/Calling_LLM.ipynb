<h1 style="font-size: 50px;">Json Output Parser</h1>

```python
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.exceptions import OutputParserException
from langchain_core.prompts import PromptTemplate
from langchain_groq import ChatGroq
```

```python
prompt ='''
From the below article , extract movie name, budget, revenue, studio name in JSON output format containing
the following keys: 'movie_title', 'Budget', 'Revenue', 'Studio'

Only return the valid JSON . No preamble

Add currency and also unit.


Artlice
=========
{article}

'''
```

```python
pt = PromptTemplate.from_template(prompt)
```

```python
article_text='''
This year's box office race has been unexpectedly fierce, with several major studios releasing high-stakes tentpoles.
Among the biggest contenders is **Crimson Horizon*, an action-thriller produced by **Vertex Motion Pictures*.
The studio poured *$95 million* into its production, banking on its explosive set pieces and a star-studded cast.
Their gamble paid off when the film earned an impressive *$310 million* worldwide within its first month.

Meanwhile, animation powerhouse *Sunstream Studios* released its long-awaited fantasy adventure *Whispering Realms*,
crafted over five years with a budget of *$120 million*. The movie resonated strongly with families and fantasy fans alike,
accumulating a global revenue of *$540 million* and becoming Sunstream's highest-grossing animated film to date.

Not to be outdone, *IronPeak Entertainment* pushed forward with its ambitious sci-fi drama *Orbital Divide*.
Despite some pre-release skepticism, the studio invested *$150 million* into the project, emphasizing groundbreaking
zero-gravity cinematography. The film surprised analysts by pulling in *$220 million*—a modest return, but still a win
for a genre known for risky box-office behavior.

Rounding out the lineup, *Grand Summit Studios* released the historical epic *Empire of Ashes*, featuring sprawling sets
and thousands of extras. With a production budget of *$200 million*, the film was a massive undertaking. Fortunately, strong
international markets helped it generate *$780 million* globally, making it one of the highest-earning historical films
ever produced.

Analysts say the diversity of genres—from sci-fi to historical epics—has contributed to one of the most financially dynamic
years the film industry has seen.



'''
```

```python
llm=ChatGroq(model="qwen/qwen3.8-27b")
```

```python
chain =pt|llm
```

```python
response= chain.invoke({'article':article_text})
```

```python
print(response.content)
```

```python
parser= JsonOutputParser()
res= parser.parse(response.content)
res
```

```python
type(res)
```

```python
res[1]['Budget']
```

```python
from sklearn.metrics.pairwise import cosine_similarity
text1="Machine Learning improves prediction accuracy"
text2="AI models help make better predictions"
text3= "Pizza is my favourite food"
text4="I am learning painting"
```

```python
# Compare Two Sentences

v1=embedder.embed_query(text1)
v2=embedder.embed_query(text2)
v3=embedder.embed_query(text3)
v4=embedder.embed_query(text4)
print("Similarity text1 <-> text2", cosine_similarity([v1], [v2]))
print("Similarity text1 <-> text3", cosine_similarity([v1], [v3]))
print("Similarity text3 <-> text2", cosine_similarity([v3], [v2]))
print("Similarity text4 <-> text3", cosine_similarity([v4], [v3]))
```
